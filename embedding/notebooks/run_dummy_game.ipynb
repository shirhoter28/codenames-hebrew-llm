# Watch a single-team dummy game

Uses the same `SingleTeamGame` loop as `scripts/run_dummy_game.py`. Dummy agents are for plumbing, not reported results.

- `guesser="dummy"`: first remaining word (often hits the assassin)
- `guesser="oracle"`: always picks Red (engine win-path check only)

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / "codenames" / "game.py").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from codenames.board import create_board
from codenames.game import SingleTeamGame
from codenames.logging_io import write_game_result
from codenames.players import DummyCodemaster, DummyGuesser, OracleGuesser
```

```python
SEED = 0
WORDPOOL = "regular"
GUESSER = "dummy"  # or "oracle"
SAVE_LOG = True

board = create_board(seed=SEED, wordpool=WORDPOOL)
codemaster = DummyCodemaster()
guesser = OracleGuesser(board.role_of) if GUESSER == "oracle" else DummyGuesser()
result = SingleTeamGame(board, codemaster, guesser, do_print=True).run()
if SAVE_LOG:
    path = write_game_result(result)
    print("Wrote", path)
```
