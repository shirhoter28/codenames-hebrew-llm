## Colored 5×5 (Codemaster view)

Red / blue / grey civilian / purple assassin. Guesser view is the same grid without colors.

```python
import pandas as pd

ROLE_COLORS = {
    "Red": "#f4c2c2",
    "Blue": "#c2d4f4",
    "Civilian": "#e8e8e8",
    "Assassin": "#d8c2f4",
}


def show_board(board, hide_roles=False):
    df = pd.DataFrame(board.grid)
    roles = pd.DataFrame(board.role_grid)
    styled = df.style.set_properties(
        **{"text-align": "center", "font-size": "18px", "width": "120px", "height": "48px"}
    ).hide(axis="index").hide(axis="columns")
    if not hide_roles:
        styled = styled.apply(
            lambda col: [
                f"background-color: {ROLE_COLORS[roles.loc[i, col.name]]}"
                for i in col.index
            ],
            axis=0,
        )
    display(styled)
    print(f"seed={board.seed}  wordpool={board.wordpool}")
    if not hide_roles:
        print("Red:", board.red)
        print("Blue:", board.blue)
        print("Civilian:", board.civilian)
        print("Assassin:", board.assassin)


show_board(board)
```

## Guesser view and other seeds

Uncomment / change the lines below and re-run.

```python
print("Guesser view (no colors):")
show_board(board, hide_roles=True)

print("\nAnother seed, dual-meaning pool:")
show_board(create_board(seed=1, wordpool="dual"))
```
